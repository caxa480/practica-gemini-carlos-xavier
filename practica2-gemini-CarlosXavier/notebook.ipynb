# Problemas Encontrados

Durante el desarrollo de la práctica se detectaron algunas limitaciones relacionadas con el uso de la API de Gemini.

### Error 503 - UNAVAILABLE

Este error apareció cuando los servidores de Gemini se encontraban bajo una carga elevada y no podían procesar temporalmente nuevas solicitudes.

Ejemplo:

503 UNAVAILABLE

### Error 429 - RESOURCE_EXHAUSTED

Este error apareció cuando se alcanzó el límite de peticiones disponible en la cuota gratuita de la API.

Ejemplo:

429 RESOURCE_EXHAUSTED

A pesar de estas limitaciones, se verificó que los scripts desarrollados funcionaban correctamente, ya que las peticiones se enviaban correctamente a Gemini y los errores eran gestionados y almacenados en los informes generados.
